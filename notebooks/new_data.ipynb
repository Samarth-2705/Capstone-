```python
import pandas as pd
import numpy as np
# load dataset
df = pd.read_csv("House_prices.csv")

print("Shape:", df.shape)
print("\nColumns:\n", df.columns)
print("\nData Types:\n", df.dtypes)
```

```python
print("\nMissing Values:\n")
print(df.isnull().sum().sort_values(ascending=False))
```

```python
df.describe().T
```

```python
df_3000 = df.iloc[:3000].copy()

print("New shape:", df_3000.shape)
df_3000.head()
```

```python
import matplotlib.pyplot as plt

# Price distribution
plt.hist(df_3000["Price"], bins=50)
plt.title("Price Distribution")
plt.show()

# Living area vs price
plt.scatter(df_3000["living area"], df_3000["Price"], alpha=0.3)
plt.xlabel("Living Area")
plt.ylabel("Price")
plt.title("Living Area vs Price")
plt.show()
```

```python
df_3000["log_price"] = np.log1p(df_3000["Price"])
```

```python
plt.hist(np.log1p(df_3000["Price"]), bins=50)
plt.title("Log Price Distribution")
plt.show()
```
